```python
from PIL import Image
import os
import json
```

```python
data = json.load(open('./assets/projects_data.json'))
```

```python
aspect_ratio = 4/3

for project in data:
  image_path = f'./assets/screenshots/{project["slug"]}.jpg'
  img = Image.open(image_path)
  width, height = img.size

  new_height = width/aspect_ratio
  if (height > new_height):
    img = img.crop((0, 0, width, new_height))

  if (width > 400):
    old_width, old_height = img.size
    new_height = int(400 / old_width * old_height)
    img = img.resize((400, new_height), Image.LANCZOS)

  filename = os.path.basename(image_path)
  name, ext = os.path.splitext(filename)

  output_path = os.path.join(os.path.dirname(image_path), f'{name}{ext}')
  img.save(output_path)
```
