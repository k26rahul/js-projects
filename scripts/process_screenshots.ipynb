```python
from PIL import Image
import os
import json
```

```python
f = open('../assets/projects_data.json')
data = json.load(f)
data[0]
```

```python
def crop_to_aspect(img, aspect_ratio: float):
  """Crop image to a given aspect ratio (width/height)."""
  width, height = img.size
  new_height = width / aspect_ratio

  if height <= new_height:
    return img, False   # skipped

  return img.crop((0, 0, width, round(new_height))), True  # cropped
```

```python
def resize_width(img, new_width: int):
  """Resize image to given width while preserving aspect ratio."""
  old_width, old_height = img.size

  if old_width <= new_width:
    return img, False   # skipped

  new_height = round(new_width / old_width * old_height)
  return img.resize((new_width, new_height), Image.LANCZOS), True  # resized
```

```python
img = Image.open("example.jpg")

img, did_cropped = crop_to_aspect(img, aspect_ratio=4/3)
img, did_resized = resize_width(img, 400)

print(f"Cropped: {did_cropped}, Resized: {did_resized}")

img.save("example_processed.jpg")
```

```python
def process_screenshots():
  """Process all project screenshots: locate each JPG or PNG by slug, crop to 4:3, resize to 400 px width, convert to RGB if needed, and save as a JPG."""

  aspect_ratio = 4/3
  max_width = 400

  for project in data:
    # try jpg first, then png
    for ext in [".jpg", ".png"]:
      image_path = f'../assets/screenshots/{project["slug"]}{ext}'
      if os.path.exists(image_path):
        break
    else:
      print(f"Skipping {project['slug']}: no JPG or PNG found")
      continue

    img = Image.open(image_path)

    # crop
    img, cropped = crop_to_aspect(img, aspect_ratio)

    # resize
    img, resized = resize_width(img, max_width)

    # convert to RGB (necessary for saving as JPG)
    converted = False
    if img.mode in ("RGBA", "LA"):
      img = img.convert("RGB")
      converted = True

    # save as JPG with slug name
    output_path = os.path.join(os.path.dirname(image_path), f'{project["slug"]}.jpg')
    img.save(output_path, "JPEG", quality=95)

    # status message
    status = []
    status.append("✅ Cropped" if cropped else "❌ Skipped cropping")
    status.append("✅ Resized" if resized else "❌ Skipped resizing")
    if converted:
      status.append("✅ Converted to RGB")

    print(f"Processed {project['slug']}.jpg ({', '.join(status)})")
```

```python
process_screenshots()
```
