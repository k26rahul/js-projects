```python
import os
import json
from datetime import datetime
```

```python
f = open('../assets/projects_data.json')
data = json.load(f)
data[0]
```

```python
for project in data:
  index_path = f"../{project['slug']}/index.html"

  # get creation time and update date
  ctime = os.path.getctime(index_path)
  project['date'] = datetime.fromtimestamp(ctime).strftime('%Y-%m-%d')
```

```python
data.sort(key=lambda project: project['date'])
```

```python
f = open('../assets/projects_data.json', 'w')
json.dump(data, f, indent=2)
f.close()
```
